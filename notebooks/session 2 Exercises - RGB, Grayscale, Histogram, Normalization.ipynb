# MILK10k — Session 2 HW: Pixels, Color Depth, Channels, Grayscale, Histograms & Normalization

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path
from PIL import Image

DATA_DIR = Path("../../milk10k_project")
```

```python
df_metadata = pd.read_csv(DATA_DIR / "metadata.csv")
gt = pd.read_csv(DATA_DIR / "supplements" / "training_gt.csv")

CLASS_COLS = ["AKIEC", "BCC", "BEN_OTH", "BKL", "DF",
              "INF", "MAL_OTH", "MEL", "NV", "SCCKA"]
gt["diagnosis_class"] = gt[CLASS_COLS].idxmax(axis=1)

df = df_metadata.merge(gt[["lesion_id", "diagnosis_class"]], on="lesion_id", how="inner")
df.head()
```

```python
image_paths = list((DATA_DIR / "images").glob("*.jpg"))
img = Image.open(image_paths[0])
img_array = np.array(img)
print(img_array.shape)
```

## Pixels & Resolution

```python
height, width = img_array.shape[0], img_array.shape[1]
print(f"Spatial resolution: {height} x {width}")
print(f"NumPy array shape: {img_array.shape}")
print(f"Array dtype: {img_array.dtype}")
```

```python
center_y, center_x = height // 2, width // 2
crop = img_array[center_y-50:center_y+50, center_x-50:center_x+50]

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(img_array)
axes[0].set_title(f"Original ({height}x{width})")
axes[0].axis("off")
axes[1].imshow(crop)
axes[1].set_title("Center Crop (100x100)")
axes[1].axis("off")
plt.tight_layout()
plt.show()
```

- **Spatial resolution (450x600):** the image is made of 450 rows and 600 columns of pixels — more pixels means finer detail can be captured about the lesion's texture and edges.
- **Shape has 3 numbers, not 2:** the third dimension is the color channels (R, G, B). A 2D shape would mean a single-channel (grayscale) image.
- **dtype uint8:** each value is an 8-bit unsigned integer, so the range is 0-255.
- **Center crop:** we lose the surrounding context (skin border, overall lesion shape/size). This would matter for diagnosis if the lesion is larger than the crop or if border irregularity (a key diagnostic feature) falls outside the cropped region.

## Color Depth

```python
def quantize(image_array, levels):
    """Reduce the number of distinct intensity levels."""
    step = 256 / levels
    return (np.floor(image_array / step) * step).astype(np.uint8)

gray_image = (0.299*img_array[:,:,0] + 0.587*img_array[:,:,1] + 0.114*img_array[:,:,2]).astype(np.uint8)

bit8 = gray_image  # 256 levels, original
bit4 = quantize(gray_image, 16)   # 4-bit
bit2 = quantize(gray_image, 4)    # 2-bit
bit1 = quantize(gray_image, 2)    # 1-bit

fig, axes = plt.subplots(1, 4, figsize=(16, 4))
titles = ["8-bit (256 levels)", "4-bit (16 levels)", "2-bit (4 levels)", "1-bit (2 levels)"]
for ax, im, title in zip(axes, [bit8, bit4, bit2, bit1], titles):
    ax.imshow(im, cmap="gray", vmin=0, vmax=255)
    ax.set_title(title)
    ax.axis("off")
plt.tight_layout()
plt.savefig("bit_depth_comparison.png")
plt.show()
```

**Question:** At what bit depth does the lesion boundary become hard to distinguish?

Typically around 2-bit (4 levels), the smooth gradient between lesion and skin collapses into large flat blocks, making the boundary blocky and ambiguous. At 1-bit it's essentially gone. This matters for a diagnostic tool because border irregularity is a key feature (the "B" in the ABCDE rule for melanoma) — losing that detail could cause a model or clinician to miss an important warning sign.

## RGB Channels

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(img_array[:, :, 0], cmap="gray")
axes[0].set_title("Red Channel (raw intensity)")
axes[0].axis("off")

axes[1].imshow(img_array[:, :, 1], cmap="gray")
axes[1].set_title("Green Channel (raw intensity)")
axes[1].axis("off")

axes[2].imshow(img_array[:, :, 2], cmap="gray")
axes[2].set_title("Blue Channel (raw intensity)")
axes[2].axis("off")

plt.tight_layout()
plt.savefig("rgb_channels_grayscale.png")
plt.show()
```

**Question:** Which channel carries the most useful signal for distinguishing a lesion?

The Red channel often shows the strongest contrast between the lesion and surrounding skin, since blood vessels and pigmented tissue tend to absorb more green/blue light while reflecting red. The Blue channel tends to be the noisiest/lowest contrast for skin lesions.

## Grayscale Conversion

```python
# Method 1: naive average
gray_naive = ((img_array[:,:,0].astype(np.float32) +
               img_array[:,:,1].astype(np.float32) +
               img_array[:,:,2].astype(np.float32)) / 3).astype(np.uint8)

# Method 2: luminosity-weighted
gray_luminosity = (0.299*img_array[:,:,0] +
                    0.587*img_array[:,:,1] +
                    0.114*img_array[:,:,2]).astype(np.uint8)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(gray_naive, cmap="gray")
axes[0].set_title("Naive Average")
axes[0].axis("off")
axes[1].imshow(gray_luminosity, cmap="gray")
axes[1].set_title("Luminosity-Weighted")
axes[1].axis("off")
plt.tight_layout()
plt.show()

max_diff = np.abs(gray_naive.astype(int) - gray_luminosity.astype(int)).max()
print(f"Max pixel difference: {max_diff}")
```

**Question:** Why do these methods disagree, and which matches human perception better?

They disagree because the naive average weights all 3 channels equally, while the luminosity formula weights green heavily (0.587) and blue lightly (0.114), matching how the human eye is most sensitive to green light and least sensitive to blue. The luminosity-weighted method better matches human perception of brightness.

### 3. Comparing grayscale histograms across diagnosis classes

```python
class_a = df[df["diagnosis_class"] == "NV"].iloc[0]
class_b = df[df["diagnosis_class"] == "MEL"].iloc[0]

img_a = np.array(Image.open(DATA_DIR / "images" / f"{class_a['isic_id']}.jpg"))
img_b = np.array(Image.open(DATA_DIR / "images" / f"{class_b['isic_id']}.jpg"))

gray_a = (0.299*img_a[:,:,0] + 0.587*img_a[:,:,1] + 0.114*img_a[:,:,2]).astype(np.uint8)
gray_b = (0.299*img_b[:,:,0] + 0.587*img_b[:,:,1] + 0.114*img_b[:,:,2]).astype(np.uint8)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].hist(gray_a.ravel(), bins=256, range=(0, 255), color="gray")
axes[0].set_title(f"Grayscale Histogram - {class_a['diagnosis_class']}")
axes[0].set_xlabel("Pixel Intensity")

axes[1].hist(gray_b.ravel(), bins=256, range=(0, 255), color="gray")
axes[1].set_title(f"Grayscale Histogram - {class_b['diagnosis_class']}")
axes[1].set_xlabel("Pixel Intensity")

plt.tight_layout()
plt.show()
```

**Question:** Do you see a visible difference in the histogram shape? What might explain it?

Yes, the two histograms typically differ in shape and where the intensity mass concentrates. This can be explained by differences in lighting conditions during capture, the size/color of the lesion itself (larger or darker lesions pull the distribution toward lower intensities), and the patient's skin tone (lighter skin shifts the distribution toward higher intensities).

## Normalization
### 4. Min-max normalization

```python
minmax_img = (img_array / 255.0).astype(np.float32)
print(f"Min-max range: [{minmax_img.min():.4f}, {minmax_img.max():.4f}]")
```

### 5. Z-score standardization

```python
zscore_img = (img_array.astype(np.float32) - img_array.mean()) / img_array.std()
print(f"Z-score mean: {zscore_img.mean():.4f}, std: {zscore_img.std():.4f}")
```

### 6. Comparing raw, min-max, and z-score histograms

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].hist(img_array.ravel(), bins=256, color="steelblue")
axes[0].set_title("Raw [0, 255]")

axes[1].hist(minmax_img.ravel(), bins=256, color="darkorange")
axes[1].set_title("Min-Max [0, 1]")

axes[2].hist(zscore_img.ravel(), bins=256, color="green")
axes[2].set_title("Z-score (mean=0, std=1)")

for ax in axes:
    ax.set_xlabel("Pixel Value")
    ax.set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

**Question:** Which of these would you use as input to a neural network, and why does normalization matter?

Z-score standardization (or min-max, depending on the architecture) is generally preferred over raw pixel values, since keeping inputs on a small, centered scale helps gradients descend more smoothly during training — unnormalized large values (0-255) can cause unstable or slow-converging gradients.
